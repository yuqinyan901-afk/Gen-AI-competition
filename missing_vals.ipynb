```python
# Load dotenv
# Get API Key
# Import libraries
# Langchain imports
import pandas as pd
import numpy as np
import os
from dotenv import load_dotenv
```

### **Helper Functions**
- `helper_function_1(df,param=None)`: helper function 1 description
- `helper_function_2(series,method=None)`: helper function 2 description
- ...

```python
def fill_mean(df):
    return df.fillna(df.mean(numeric_only=True))

def fill_median(df):
    return df.fillna(df.median(numeric_only=True))

def fill_mode(df):
    return df.fillna(df.mode().iloc[0])

def drop_missing(df):
    return df.dropna()

def forward_fill(df):
    return df.fillna(method="ffill")

def backward_fill(df):
    return df.fillna(method="bfill")

def route_missing_strategy(user_query):
    query = user_query.lower()

    if "mean" in query:
        return "mean"
    elif "median" in query:
        return "median"
    elif "mode" in query:
        return "mode"
    elif "drop" in query:
        return "drop"
    elif "forward" in query or "ffill" in query:
        return "ffill"
    elif "backward" in query or "bfill" in query:
        return "bfill"
    else:
        return "mean"  # default
```

# **MAIN FEATURE FUNCTION**

```python
def missing_vals(df, user_query):
    print(f"User request: {user_query}")

    strategy = route_missing_strategy(user_query)
    print(f"Selected strategy: {strategy}")

    if strategy == "mean":
        df = fill_mean(df)
    elif strategy == "median":
        df = fill_median(df)
    elif strategy == "mode":
        df = fill_mode(df)
    elif strategy == "drop":
        df = drop_missing(df)
    elif strategy == "ffill":
        df = forward_fill(df)
    elif strategy == "bfill":
        df = backward_fill(df)

    print("Missing values handled successfully.")
    return df
```
